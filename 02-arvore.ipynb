```python
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import plot_tree

import pandas

from sklearn.tree import DecisionTreeClassifier
from sklearn.inspection import DecisionBoundaryDisplay
```

## Dados das pétalas

```python
dataset = load_iris()
```

```python
x = pandas.DataFrame(dataset['data'], columns= dataset['feature_names'])
y = pandas.Series(dataset['target'], name= 'target')
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

## Arvore de Decisão


```python
arvore = DecisionTreeClassifier (
    max_depth=2,
    random_state=42
)
```

```python
arvore.fit(x_train, y_train)
```

```python
print("Acurácia da Arvore Treino :", arvore.score(x_train, y_train))
print("Acurácia da Arvore Teste :", arvore.score(x_test, y_test))
```

```python
plt.figure(figsize=(14,8))
```

```python
plot_tree (
    arvore, 
    feature_names=x.columns,
    class_names=dataset.target_names,
    filled=True     
)
```

```python
plt.show()
```

# Grafico dos Dados por Pares

```python
X = dataset.data[:, :2]
subarvore = DecisionTreeClassifier(max_depth=2, random_state=42)
subarvore.fit(X, dataset.target)
disp = DecisionBoundaryDisplay.from_estimator(
    subarvore, X, response_method="predict",
    xlabel=dataset.feature_names[0], ylabel=dataset.feature_names[1],
    alpha=0.7
)
cmap = ListedColormap(disp.multiclass_colors_)
disp.ax_.scatter(X[:, 0], X[:, 1], c=dataset.target, edgecolor="k", cmap=cmap)

plt.show()
```

```python
for pairidx, pair in enumerate([[0, 1], [0, 2], [0, 3], [1, 2], [1, 3], [2, 3]]):
    # We only take the two corresponding features
    X = dataset.data[:, pair]
    y = dataset.target

    # Train
    clf = DecisionTreeClassifier().fit(X, y)

    # Plot the decision boundary
    ax = plt.subplot(2, 3, pairidx + 1)
    plt.tight_layout(h_pad=0.5, w_pad=0.5, pad=2.5)
    disp = DecisionBoundaryDisplay.from_estimator(
        clf,
        X,
        response_method="predict",
        ax=ax,
        xlabel=dataset.feature_names[pair[0]],
        ylabel=dataset.feature_names[pair[1]],
        alpha=0.5,
    )

    # Plot the training points
    scatter = disp.ax_.scatter(
        X[:, 0],
        X[:, 1],
        c=y,
        cmap=ListedColormap(disp.multiclass_colors_),
        edgecolor="black",
        s=15,
    )

plt.suptitle("Decision surface of decision trees trained on pairs of features")
plt.figlegend(
    scatter.legend_elements()[0],
    dataset.target_names,
    loc="lower center",
    ncols=len(dataset.target_names),
)
_ = plt.axis("tight")
```
